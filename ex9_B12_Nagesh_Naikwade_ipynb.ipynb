1. What is a Data Science pipeline? Explain its key stages with examples.

A Data Science pipeline is a structured sequence of steps used to process raw data and transform it into meaningful insights or predictions. It begins with data collection, where raw data is gathered from sources such as CSV files, databases, or APIs. The next stage is data cleaning, which involves handling missing values, removing duplicates, and correcting inconsistencies. After cleaning, Exploratory Data Analysis (EDA) is performed to understand patterns, trends, and relationships within the data using visualizations and statistics. Feature engineering is then applied to create new meaningful variables, such as calculating total marks from subject-wise scores. The processed data is then used for model building, where machine learning algorithms like Linear Regression are trained. Finally, the model is evaluated using metrics such as accuracy or error rates to determine its performance. For example, a pipeline can be used to predict student performance based on study hours and attendance.


2. What is Exploratory Data Analysis (EDA)? How do univariate and bivariate analyses help in understanding the dataset?

Exploratory Data Analysis (EDA) is the process of analyzing and visualizing data to better understand its structure, patterns, and relationships before applying machine learning models. It helps in identifying trends, detecting anomalies, and making informed decisions about preprocessing. Univariate analysis focuses on a single variable at a time and helps in understanding its distribution, central tendency, and spread using tools like histograms or box plots. For example, analyzing the distribution of student marks helps identify whether the data is skewed or normally distributed. On the other hand, bivariate analysis examines the relationship between two variables, such as study hours and marks, using scatter plots or correlation matrices. This helps in identifying whether variables are positively or negatively related, which is important for building predictive models.


3. Differentiate between Label Encoding and One-Hot Encoding. When should each be used?

Label Encoding and One-Hot Encoding are techniques used to convert categorical data into numerical form so that machine learning models can process it. Label Encoding assigns a unique integer value to each category, such as assigning 1 to “Red,” 2 to “Blue,” and 3 to “Green.” This method is simple and memory-efficient but may introduce unintended ordinal relationships between categories. In contrast, One-Hot Encoding creates separate binary columns for each category, where a value of 1 indicates the presence of a category and 0 indicates its absence. For example, a “Color” column would be split into multiple columns like “Red,” “Blue,” and “Green.” Label Encoding is best used when the categorical data has a natural order, such as “Low,” “Medium,” and “High,” while One-Hot Encoding is preferred for nominal data where no order exists, such as colors or cities.


4. Why is feature scaling required? Compare normalization and standardization.

Feature scaling is required in machine learning to ensure that all input features contribute equally to the model and to improve the efficiency and accuracy of algorithms. Many machine learning algorithms, especially those based on distance calculations like K-Nearest Neighbors or gradient-based methods, are sensitive to the scale of data. If one feature has a much larger range than others, it can dominate the model’s learning process. Normalization and standardization are two common methods of feature scaling. Normalization rescales the data to a fixed range, usually between 0 and 1, using the minimum and maximum values of the feature. It is useful when the data does not follow a normal distribution and is commonly used in neural networks. Standardization, on the other hand, transforms the data so that it has a mean of 0 and a standard deviation of 1. It is more suitable for algorithms that assume normally distributed data, such as Linear Regression. Both techniques help improve model performance and convergence speed.



5. What is feature engineering? Explain how creating new features can improve model performance

Feature engineering is the process of creating new input features or modifying existing ones to improve the performance of machine learning models. It involves transforming raw data into more meaningful representations that better capture the underlying patterns. For example, instead of using separate columns for math and science marks, a new feature called total marks can be created, which may better represent overall performance. Similarly, extracting the day, month, or year from a date column can provide additional insights. Feature engineering can also include techniques like encoding categorical variables, scaling numerical data, and combining multiple features. By providing more relevant and informative inputs, feature engineering helps the model learn more effectively, reduces noise, and often leads to higher accuracy and better predictions.

```python
# Step 1: Import Libraries
import pandas as pd
import numpy as np
from google.colab import files
import zipfile
import io

# Upload the file if not already uploaded
if 'uploaded' not in globals():
    uploaded = files.upload()

# Check if 'archive.zip' was uploaded and extract its contents
if 'archive.zip' in uploaded:
    with zipfile.ZipFile(io.BytesIO(uploaded['archive.zip']), 'r') as z:
        z.extractall() # Extract all contents of the zip file to the current directory

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Step 2: Load Dataset
df = pd.read_csv("job_salary_prediction_dataset.csv")

# Step 3: Preview Data
print(df.head())
print("\nColumns:", df.columns)

# Step 4: Handle Missing Values
df = df.dropna()

# Step 5: Encode Categorical Columns
le = LabelEncoder()

for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = le.fit_transform(df[col])

# Step 6: Set Target Column (Salary)
# Try common names safely
if 'salary' in df.columns:
    target = 'salary'
elif 'Salary' in df.columns:
    target = 'Salary'
else:
    target = df.columns[-1]   # fallback

print("Target Column:", target)

# Step 7: Split Features & Target
X = df.drop(target, axis=1)
y = df[target]

# Step 8: Ensure Numeric Data Only
X = X.select_dtypes(include=[np.number])

# Step 9: Feature Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 10: Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# Step 11: Train Model
model = LinearRegression()
model.fit(X_train, y_train)

# Step 12: Predictions
y_pred = model.predict(X_test)

# Step 13: Evaluation
print("\nModel Performance:")
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
```
